```python
import os
import whisper
```

```python
model=whisper.load_model('tiny')
```

```python
def speechToText(audio):
    text=model.transcribe(audio)
    print(text)
    return(text)
```

```python
audiofilepath='C:/test/audio/'#the path of audio(need to modify)
textfilepath='C:/test/text_trans/'#the path of audio(need to modify)
audiofilelist=os.listdir(audiofilepath)
audiofiletime=[]
for i in range(len(audiofilelist)):
    filetime=os.path.getmtime(audiofilepath+audiofilelist[i])
    audiofiletime.append(filetime)
maxVal=max(audiofiletime)
maxInd=audiofiletime.index(maxVal)
latestfilename=audiofilelist[maxInd]
```

```python
Output=speechToText(audiofilepath+latestfilename)
```

```python
with open(textfilepath+latestfilename+'.txt','w') as f:
    f.write(Output['text'])
```
